**Read and Show Image**

```python
import cv2

image = cv2.imread( "images/images (1).jpg")
if image is None:
    print("Image not found!")
else:
    print("Image loaded successfully!")
    cv2.imshow("My Image", image)
    cv2.waitKey(0) #Wait for a key press to close the window
    cv2.destroyAllWindows()
```

**Show Multiple Images**

```python
import cv2

# Read images
image1 = cv2.imread("images/images (1).jpg")
image2 = cv2.imread("images\\beautiful-fall-nature-scenery-free-image.jpeg")
image3 = cv2.imread("images\\istockphoto-517188688-1024x1024.jpg")
cv2.imshow("Image 1", image1)
cv2.imshow("Image 2", image2)
cv2.imshow("Image 3", image3)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

**Slideshow**

```python
import cv2
images = [
    "images\\images (1).jpg",
    "images\\beautiful-fall-nature-scenery-free-image.jpeg",
    "images\\istockphoto-517188688-1024x1024.jpg"
]
for path in images:
    image = cv2.imread(path)
    if image is None:
        print("Image not found:", path)
        continue
    cv2.imshow("Slideshow", image)
    key = cv2.waitKey(2000) # 2 seconds
    if key == 27:
        break
cv2.destroyAllWindows()
```
