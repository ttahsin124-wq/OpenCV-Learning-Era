# Draw Polygons on Image

A polygon is a closed shape made from multiple points.

OpenCV provides:

- `cv2.polylines()` → draws the outline of a polygon
- `cv2.fillPoly()` → draws a filled polygon

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
image = cv2.imread("images/360_F_383258331_D8imaEMl8Q3lf7EKU2Pi78Cn0R7KkW9o.jpg")
points = np.array([
    [150, 100],
    [250, 50],
    [350, 100],
    [300, 200],
    [200, 200]
], np.int32)
points = points.reshape((-1, 1, 2))
cv2.polylines(
    image,
    [points],
    True,          
    (155, 0, 0),    
    3                  
)
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(10, 6))
plt.imshow(image_rgb)
plt.axis("off")
plt.show()
```

# Arithmetic Operations on Images

OpenCV allows us to perform arithmetic operations on images.

In this section, we will learn:

- `cv2.add()` → adds two images
- `cv2.subtract()` → subtracts one image from another

For these operations, both images should have the same size.

Image Addition

```python
import cv2
import matplotlib.pyplot as plt

# Read two images
image1 = cv2.imread("images\images (2).jpg")
image2 = cv2.imread("images\images (3).jpg")

# Check if images were loaded
if image1 is None or image2 is None:
    print("One or both images were not found!")
else:
    # Resize image2 to match image1
    image2 = cv2.resize(image2, (image1.shape[1], image1.shape[0]))

    # Add the two images
    added_image = cv2.add(image1, image2)

    # Convert BGR to RGB
    image1_rgb = cv2.cvtColor(image1, cv2.COLOR_BGR2RGB)
    image2_rgb = cv2.cvtColor(image2, cv2.COLOR_BGR2RGB)
    added_rgb = cv2.cvtColor(added_image, cv2.COLOR_BGR2RGB)

    # Display
    plt.figure(figsize=(15, 5))

    plt.subplot(1, 3, 1)
    plt.imshow(image1_rgb)
    plt.title("Image 1")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(image2_rgb)
    plt.title("Image 2")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(added_rgb)
    plt.title("Added Image")
    plt.axis("off")

    plt.show()
```

## Image Subtraction

`cv2.subtract()` subtracts the pixel values of one image from another.

Syntax:

cv2.subtract(image1, image2)

```python
import cv2
import matplotlib.pyplot as plt


image1 = cv2.imread("images\images (2).jpg")
image2 = cv2.imread("images\images (3).jpg")

if image1 is None or image2 is None:
    print("One or both images were not found!")
else:
    image2 = cv2.resize(image2, (image1.shape[1], image1.shape[0]))

    # Subtract image2 from image1
    subtracted_image = cv2.subtract(image1, image2)

    # Convert BGR to RGB
    image1_rgb = cv2.cvtColor(image1, cv2.COLOR_BGR2RGB)
    image2_rgb = cv2.cvtColor(image2, cv2.COLOR_BGR2RGB)
    subtracted_rgb = cv2.cvtColor(subtracted_image, cv2.COLOR_BGR2RGB)

    plt.figure(figsize=(15, 5))

    plt.subplot(1, 3, 1)
    plt.imshow(image1_rgb)
    plt.title("Image 1")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(image2_rgb)
    plt.title("Image 2")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(subtracted_rgb)
    plt.title("Subtracted Image")
    plt.axis("off")

    plt.show()
```

# Bitwise Operations on Binary Images

Bitwise operations are used to combine or manipulate binary images.

A binary image mainly contains two pixel values:

- 0   → Black
- 255 → White

OpenCV provides:

- `cv2.bitwise_and()` → AND operation
- `cv2.bitwise_or()`  → OR operation
- `cv2.bitwise_xor()` → XOR operation
- `cv2.bitwise_not()` → NOT operation

Create Two Binary Images

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
image1 = np.zeros((400, 400), dtype=np.uint8)
image2 = np.zeros((400, 400), dtype=np.uint8)
cv2.rectangle(   # white rectangle
    image1,
    (50, 100),                 
    (250, 300),
    255,
    -1
)

# Draw white circle on image2
cv2.circle(
    image2,
    (250, 200),
    100,
    255,
    -1
)

# Display 
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(image1, cmap="gray")
plt.title("Binary Image 1")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(image2, cmap="gray")
plt.title("Binary Image 2")
plt.axis("off")

plt.show()
```

## Bitwise AND

`cv2.bitwise_and()` keeps a pixel white only when the
corresponding pixels in BOTH images are white.

Syntax:

cv2.bitwise_and(image1, image2)

```python
result_and = cv2.bitwise_and(image1, image2)

# Display
plt.figure(figsize=(5, 5))
plt.imshow(result_and, cmap="gray")
plt.title("Bitwise AND")
plt.axis("off")
plt.show()
```

## Easy Way to Understand And
Black AND Black = Black
Black AND White = Black
White AND Black = Black
White AND White = White
Image 1: Rectangle
Image 2: Circle

        Rectangle
       ┌───────────┐
       │      ◯    │
       │    ◉      │
       │      ◯    │
       └───────────┘

          ↓ AND

Only the overlapping part remains.

## Bitwise OR

`cv2.bitwise_or()` makes a pixel white when at least
one of the corresponding pixels is white.

Syntax:

cv2.bitwise_or(image1, image2)

```python
# Bitwise OR
result_or = cv2.bitwise_or(image1, image2)
plt.figure(figsize=(5, 5))
plt.imshow(result_or, cmap="gray")
plt.title("Bitwise OR")
plt.axis("off")
plt.show()
```

## OR table
Black OR Black = Black
Black OR White = White
White OR Black = White
White OR White = White
So or gives the Combined area of both shapes!

## Bitwise XOR

`cv2.bitwise_xor()` makes a pixel white when the two
corresponding pixels are different.

Syntax:

cv2.bitwise_xor(image1, image2)

```python
result_xor = cv2.bitwise_xor(image1, image2)
plt.figure(figsize=(5, 5))
plt.imshow(result_xor, cmap="gray")
plt.title("Bitwise XOR")
plt.axis("off")
plt.show()
```

## XOR Table
Black XOR Black = Black
Black XOR White = White
White XOR Black = White
White XOR White = Black

## The overlapping area becomes black because both pixels are white.

## Bitwise NOT

`cv2.bitwise_not()` reverses the pixel values.

- Black → White
- White → Black

Syntax:

cv2.bitwise_not(image)

```python
result_not = cv2.bitwise_not(image1)
plt.figure(figsize=(5, 5))
plt.imshow(result_not, cmap="gray")
plt.title("Bitwise NOT")
plt.axis("off")
plt.show()
```

## See All Operations Together

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
image1 = np.zeros((400, 400), dtype=np.uint8)
image2 = np.zeros((400, 400), dtype=np.uint8)
cv2.rectangle(image1, (50, 100), (250, 300), 255, -1)
cv2.circle(image2, (250, 200), 100, 255, -1)
result_and = cv2.bitwise_and(image1, image2)
result_or = cv2.bitwise_or(image1, image2)
result_xor = cv2.bitwise_xor(image1, image2)
result_not = cv2.bitwise_not(image1)
plt.figure(figsize=(12, 8))
plt.subplot(2, 3, 1)
plt.imshow(image1, cmap="gray")
plt.title("Image 1")
plt.axis("off")
plt.subplot(2, 3, 2)
plt.imshow(image2, cmap="gray")
plt.title("Image 2")
plt.axis("off")
plt.subplot(2, 3, 3)
plt.imshow(result_and, cmap="gray")
plt.title("AND")
plt.axis("off")
plt.subplot(2, 3, 4)
plt.imshow(result_or, cmap="gray")
plt.title("OR")
plt.axis("off")
plt.subplot(2, 3, 5)
plt.imshow(result_xor, cmap="gray")
plt.title("XOR")
plt.axis("off")
plt.subplot(2, 3, 6)
plt.imshow(result_not, cmap="gray")
plt.title("NOT")
plt.axis("off")
plt.tight_layout()
plt.show()
```

# Edge Detection in an Image

Edge detection is a technique used to identify the boundaries of objects
in an image.

One of the most commonly used edge detection methods in OpenCV is
the **Canny Edge Detector**.

OpenCV function:

`cv2.Canny()`

```python
import cv2
import matplotlib.pyplot as plt

# Read the image
image = cv2.imread(r"images\images (2).jpg")
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Detect edges
edges = cv2.Canny(gray, 100, 200)
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(image_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(edges, cmap="gray")
plt.title("Edge Detected Image")
plt.axis("off")

plt.show()
```

# Image Thresholding

Thresholding is a technique used to separate objects or regions
from the background of an image.

It usually works on a grayscale image.

Basic idea:

- Pixel value below the threshold → 0 (Black)
- Pixel value above the threshold → 255 (White)

OpenCV provides the `cv2.threshold()` function for basic thresholding.

```python
import cv2
import matplotlib.pyplot as plt

# Read the image
image = cv2.imread(r"images\images (2).jpg")

# Convert to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Display grayscale image
plt.figure(figsize=(6, 5))
plt.imshow(gray, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")
plt.show()
```

## Binary Thresholding

In binary thresholding:

- If pixel value > threshold → 255
- If pixel value <= threshold → 0

Syntax:

`cv2.threshold(image, threshold_value, max_value, threshold_type)`

```python
threshold_value = 127

ret, binary = cv2.threshold(
    gray,
    threshold_value,
    255,
    cv2.THRESH_BINARY
)
plt.figure(figsize=(6, 5))
plt.imshow(binary, cmap="gray")
plt.title("Binary Threshold")
plt.axis("off")
plt.show()
```

## Comparing Different Threshhold methods!

```python
import cv2
import matplotlib.pyplot as plt
# Read image
image = cv2.imread(r"images\images (2).jpg")
# Convert to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
# Binary
_, binary = cv2.threshold(
    gray, 127, 255, cv2.THRESH_BINARY
)
# Binary inverse
_, binary_inv = cv2.threshold(
    gray, 127, 255, cv2.THRESH_BINARY_INV
)
# Adaptive mean
adaptive_mean = cv2.adaptiveThreshold(
    gray,
    255,
    cv2.ADAPTIVE_THRESH_MEAN_C,
    cv2.THRESH_BINARY,
    11,
    2
)
# Adaptive Gaussian
adaptive_gaussian = cv2.adaptiveThreshold(
    gray,
    255,
    cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
    cv2.THRESH_BINARY,
    11,
    2
)
# Otsu
_, otsu = cv2.threshold(
    gray,
    0,
    255,
    cv2.THRESH_BINARY + cv2.THRESH_OTSU
)
plt.figure(figsize=(15, 10))

plt.subplot(2, 3, 1)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale")
plt.axis("off")

plt.subplot(2, 3, 2)
plt.imshow(binary, cmap="gray")
plt.title("Binary")
plt.axis("off")

plt.subplot(2, 3, 3)
plt.imshow(binary_inv, cmap="gray")
plt.title("Binary Inverse")
plt.axis("off")

plt.subplot(2, 3, 4)
plt.imshow(adaptive_mean, cmap="gray")
plt.title("Adaptive Mean")
plt.axis("off")

plt.subplot(2, 3, 5)
plt.imshow(adaptive_gaussian, cmap="gray")
plt.title("Adaptive Gaussian")
plt.axis("off")

plt.subplot(2, 3, 6)
plt.imshow(otsu, cmap="gray")
plt.title("Otsu")
plt.axis("off")

plt.tight_layout()
plt.show()
```

## Image Scaling

Image scaling means changing the size of an image.

We use:

`cv2.resize()`

Syntax:

`cv2.resize(image, new_size)`

```python
import cv2
import matplotlib.pyplot as plt
image = cv2.imread(r"images\images (2).jpg")
resized_image = cv2.resize(image, (300, 200))
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
resized_rgb = cv2.cvtColor(resized_image, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(image_rgb)
plt.title("Original Image")
plt.axis("off")
plt.subplot(1, 2, 2)
plt.imshow(resized_rgb)
plt.title("Resized Image")
plt.axis("off")
plt.show()
```

Instead of manually specifying width and height, you can scale the image using a factor.

```python
import cv2
import matplotlib.pyplot as plt
image = cv2.imread(r"images\images (2).jpg")

# Scale by 50%
scaled_image = cv2.resize(
    image,
    None,
    fx=0.5,
    fy=0.5
)
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
scaled_rgb = cv2.cvtColor(scaled_image, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(image_rgb)
plt.title("Original")
plt.axis("off")
plt.subplot(1, 2, 2)
plt.imshow(scaled_rgb)
plt.title("50% Scaled")
plt.axis("off")

plt.show()
```

## Interpolation

```python
import cv2
import matplotlib.pyplot as plt

image = cv2.imread(r"images\images (2).jpg")
nearest = cv2.resize(
    image,
    (400, 300),
    interpolation=cv2.INTER_NEAREST
)

linear = cv2.resize(
    image,
    (400, 300),
    interpolation=cv2.INTER_LINEAR
)

cubic = cv2.resize(
    image,
    (400, 300),
    interpolation=cv2.INTER_CUBIC
)
nearest = cv2.cvtColor(nearest, cv2.COLOR_BGR2RGB)
linear = cv2.cvtColor(linear, cv2.COLOR_BGR2RGB)
cubic = cv2.cvtColor(cubic, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(nearest)
plt.title("Nearest")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(linear)
plt.title("Linear")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(cubic)
plt.title("Cubic")
plt.axis("off")

plt.show()
```

## Built-in Image Rotation

OpenCV provides `cv2.rotate()` for common rotations.

- `ROTATE_90_CLOCKWISE`
- `ROTATE_180`
- `ROTATE_90_COUNTERCLOCKWISE`

```python
import cv2
import matplotlib.pyplot as plt

# Read image
image = cv2.imread(r"images\images (2).jpg")

# Rotate
rotate_90 = cv2.rotate(
    image,
    cv2.ROTATE_90_CLOCKWISE
)

rotate_180 = cv2.rotate(
    image,
    cv2.ROTATE_180
)

rotate_270 = cv2.rotate(
    image,
    cv2.ROTATE_90_COUNTERCLOCKWISE
)

# Convert BGR to RGB
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
rotate_90 = cv2.cvtColor(rotate_90, cv2.COLOR_BGR2RGB)
rotate_180 = cv2.cvtColor(rotate_180, cv2.COLOR_BGR2RGB)
rotate_270 = cv2.cvtColor(rotate_270, cv2.COLOR_BGR2RGB)

# Display
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(image_rgb)
plt.title("Original")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(rotate_90)
plt.title("90° Clockwise")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(rotate_180)
plt.title("180°")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(rotate_270)
plt.title("270° / 90° Counterclockwise")
plt.axis("off")

plt.show()
```

## Scaling + Rotation Together

```python
import cv2
import matplotlib.pyplot as plt
image = cv2.imread(r"images\images (2).jpg")
scaled = cv2.resize(
    image,
    None,
    fx=0.7,
    fy=0.7
)

height, width = scaled.shape[:2]
center = (width // 2, height // 2)

# Rotation matrix
matrix = cv2.getRotationMatrix2D(
    center,
    30,
    1.0
)

result = cv2.warpAffine(
    scaled,
    matrix,
    (width, height)
)

result_rgb = cv2.cvtColor(result, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(8, 6))
plt.imshow(result_rgb)
plt.title("Scaled + Rotated Image")
plt.axis("off")
plt.show()
```

# Image Blurring using OpenCV

Image blurring is used to smooth an image and reduce noise.
Blurring works by combining information from neighboring pixels.
OpenCV provides several blur functions:

- `cv2.blur()` → Average blur
- `cv2.GaussianBlur()` → Gaussian blur
- `cv2.medianBlur()` → Median blur
- `cv2.bilateralFilter()` → Preserves edges while reducing noise

## Average Blur

`cv2.blur()` applies an average filter to the image.

Syntax:

`cv2.blur(image, kernel_size)`

A larger kernel produces more blurring.

```python
import cv2
import matplotlib.pyplot as plt
image = cv2.imread(r"images\images (2).jpg")
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(7, 5))
plt.imshow(image_rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
#  average blur
blurred = cv2.blur(image, (5, 5))
blurred_rgb = cv2.cvtColor(blurred, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(image_rgb)
plt.title("Original Image")
plt.axis("off")
plt.subplot(1, 2, 2)
plt.imshow(blurred_rgb)
plt.title("Average Blur")
plt.axis("off")

plt.show()
```

## Gaussian Blur

`cv2.GaussianBlur()` applies a Gaussian filter to the image.

Syntax:

`cv2.GaussianBlur(image, kernel_size, sigmaX)`

```python
# Gaussian blur
gaussian = cv2.GaussianBlur(
    image,
    (5, 5),
    0
)
gaussian_rgb = cv2.cvtColor(gaussian, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(image_rgb)
plt.title("Original Image")
plt.axis("off")
plt.subplot(1, 2, 2)
plt.imshow(gaussian_rgb)
plt.title("Gaussian Blur")
plt.axis("off")

plt.show()
```

## Median Blur

`cv2.medianBlur()` applies a median filter.

Syntax:

`cv2.medianBlur(image, kernel_size)`

The kernel size must be an odd number greater than 1.

```python
#median blur
median = cv2.medianBlur(
    image,
    5
)
median_rgb = cv2.cvtColor(median, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(image_rgb)
plt.title("Original Image")
plt.axis("off")
plt.subplot(1, 2, 2)
plt.imshow(median_rgb)
plt.title("Median Blur")
plt.axis("off")
plt.show()
```

## Bilateral Filter

`cv2.bilateralFilter()` reduces noise while preserving edges.

Syntax:

`cv2.bilateralFilter(image, diameter, sigmaColor, sigmaSpace)`

```python
#bilateral filter
bilateral = cv2.bilateralFilter(
    image,
    9,
    75,
    75
)
bilateral_rgb = cv2.cvtColor(bilateral, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(image_rgb)
plt.title("Original Image")
plt.axis("off")
plt.subplot(1, 2, 2)
plt.imshow(bilateral_rgb)
plt.title("Bilateral Filter")
plt.axis("off")

plt.show()
```

## Compare all blurring methods

```python
import cv2
import matplotlib.pyplot as plt
image = cv2.imread(r"images\images (2).jpg")
original_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
average = cv2.blur(image, (5, 5))
gaussian = cv2.GaussianBlur(image, (5, 5), 0)
median = cv2.medianBlur(image, 5)
bilateral = cv2.bilateralFilter(image, 9, 75, 75)
average_rgb = cv2.cvtColor(average, cv2.COLOR_BGR2RGB)
gaussian_rgb = cv2.cvtColor(gaussian, cv2.COLOR_BGR2RGB)
median_rgb = cv2.cvtColor(median, cv2.COLOR_BGR2RGB)
bilateral_rgb = cv2.cvtColor(bilateral, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(15, 10))
plt.subplot(2, 3, 1)
plt.imshow(original_rgb)
plt.title("Original")
plt.axis("off")
plt.subplot(2, 3, 2)
plt.imshow(average_rgb)
plt.title("Average Blur")
plt.axis("off")
plt.subplot(2, 3, 3)
plt.imshow(gaussian_rgb)
plt.title("Gaussian Blur")
plt.axis("off")
plt.subplot(2, 3, 4)
plt.imshow(median_rgb)
plt.title("Median Blur")
plt.axis("off")
plt.subplot(2, 3, 5)
plt.imshow(bilateral_rgb)
plt.title("Bilateral Filter")
plt.axis("off")
plt.tight_layout()
plt.show()
```

# Make Border using OpenCV

`cv2.copyMakeBorder()` adds extra pixels around the edges of an image.

Syntax:

`cv2.copyMakeBorder(image, top, bottom, left, right, borderType)`

Parameters:

- `image` → input image
- `top` → border size at the top
- `bottom` → border size at the bottom
- `left` → border size on the left
- `right` → border size on the right
- `borderType` → type of border

```python
import cv2
import matplotlib.pyplot as plt #Constant Border
image = cv2.imread(r"images\images (2).jpg")
bordered_image = cv2.copyMakeBorder(
    image,
    50,                 # top
    50,                 # bottom
    50,                 # left
    50,                 # right
    cv2.BORDER_CONSTANT,
    value=(0, 0, 255)   # Red color (BGR)
)
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
bordered_rgb = cv2.cvtColor(bordered_image, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(image_rgb)
plt.title("Original Image")
plt.axis("off")
plt.subplot(1, 2, 2)
plt.imshow(bordered_rgb)
plt.title("Image with Border")
plt.axis("off")
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
image = cv2.imread(r"images\images (2).jpg")
bordered_image = cv2.copyMakeBorder(   # different border sizes
    image,
    20,                 # top
    40,                 # bottom
    60,                 # left
    80,                 # right
    cv2.BORDER_CONSTANT,
    value=(255, 0, 0)   # Blue
)
bordered_rgb = cv2.cvtColor(bordered_image, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(9, 6))
plt.imshow(bordered_rgb)
plt.title("Different Border Sizes")
plt.axis("off")
plt.show()
```

## Different Border Types

OpenCV provides several ways to fill the new border area:

- `BORDER_CONSTANT`
- `BORDER_REPLICATE`
- `BORDER_REFLECT`
- `BORDER_REFLECT_101`
- `BORDER_WRAP`

```python
import cv2
import matplotlib.pyplot as plt

image = cv2.imread(r"images\images (2).jpg")
constant = cv2.copyMakeBorder(
    image, 40, 40, 40, 40,
    cv2.BORDER_CONSTANT,
    value=(0, 255, 0)
)
replicate = cv2.copyMakeBorder(
    image, 40, 40, 40, 40,
    cv2.BORDER_REPLICATE
)
reflect = cv2.copyMakeBorder(
    image, 40, 40, 40, 40,
    cv2.BORDER_REFLECT
)
reflect101 = cv2.copyMakeBorder(
    image, 40, 40, 40, 40,
    cv2.BORDER_REFLECT_101
)

wrap = cv2.copyMakeBorder(
    image, 40, 40, 40, 40,
    cv2.BORDER_WRAP
)
constant = cv2.cvtColor(constant, cv2.COLOR_BGR2RGB)
replicate = cv2.cvtColor(replicate, cv2.COLOR_BGR2RGB)
reflect = cv2.cvtColor(reflect, cv2.COLOR_BGR2RGB)
reflect101 = cv2.cvtColor(reflect101, cv2.COLOR_BGR2RGB)
wrap = cv2.cvtColor(wrap, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(15, 10))
plt.subplot(2, 3, 1)
plt.imshow(constant)
plt.title("Constant")
plt.axis("off")
plt.subplot(2, 3, 2)
plt.imshow(replicate)
plt.title("Replicate")
plt.axis("off")
plt.subplot(2, 3, 3)
plt.imshow(reflect)
plt.title("Reflect")
plt.axis("off")
plt.subplot(2, 3, 4)
plt.imshow(reflect101)
plt.title("Reflect 101")
plt.axis("off")
plt.subplot(2, 3, 5)
plt.imshow(wrap)
plt.title("Wrap")
plt.axis("off")
plt.tight_layout()
plt.show()
```

# Play a Video using OpenCV

A video is basically a sequence of images called **frames**.

OpenCV can:
- Open a video file
- Read each frame
- Display the frames one by one
- Close the video when it ends or when we press a key

Important functions:

- `cv2.VideoCapture()` → opens a video
- `cap.read()` → reads one frame
- `cv2.imshow()` → displays the frame
- `cv2.waitKey()` → controls playback speed / keyboard input
- `cap.release()` → releases the video

```python
import cv2
cap = cv2.VideoCapture(r"images\Pothbondhu_Promotional_video (1).mp4")
if not cap.isOpened():
    print("Could not open the video!")
else:
    print("Video opened successfully!")
while True:
    ret, frame = cap.read()
    if not ret:
        print("Video ended.")
        break
    cv2.imshow("Video", frame)

    if cv2.waitKey(25) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
