# Text Over an Image
We can add text to an image using `cv2.putText()`.
Syntax:
cv2.putText(image, text, position, font, font_size, color, thickness)

```python
import cv2
image = cv2.imread("images\\images (1).jpg")
cv2.putText(
    image,
    "OpenCV Learning Era",
    (50, 50),
    cv2.FONT_HERSHEY_SIMPLEX,
    1,
    (0, 255, 0),
    2
)
cv2.imshow("Image with Text", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# Draw Line and Rectangle on Image

OpenCV provides drawing functions that allow us to draw shapes on images.

- `cv2.line()` → draws a line
- `cv2.rectangle()` → draws a rectangle

Both functions modify the image directly.

```python
import cv2
import matplotlib.pyplot as plt

image = cv2.imread("images/360_F_383258331_D8imaEMl8Q3lf7EKU2Pi78Cn0R7KkW9o.jpg")  
cv2.line(
    image,
    (50, 100),
    (300, 100),
    (0, 255, 0),
    3
)

# Convert BGR to RGB for displaying with matplotlib
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
plt.imshow(image_rgb)
plt.axis("off")
plt.show()
```

# Draw Polygons on Image

A polygon is a closed shape made from multiple points.

OpenCV provides:

- `cv2.polylines()` → draws the outline of a polygon
- `cv2.fillPoly()` → draws a filled polygon

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
image = cv2.imread("images/360_F_383258331_D8imaEMl8Q3lf7EKU2Pi78Cn0R7KkW9o.jpg")
points = np.array([
    [150, 100],
    [250, 50],
    [350, 100],
    [300, 200],
    [200, 200]
], np.int32)
points = points.reshape((-1, 1, 2))
cv2.polylines(
    image,
    [points],
    True,          
    (155, 0, 0),    
    3                  
)
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(10, 6))
plt.imshow(image_rgb)
plt.axis("off")
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
image = cv2.imread("images/360_F_383258331_D8imaEMl8Q3lf7EKU2Pi78Cn0R7KkW9o.jpg")
points = np.array([
    [150, 100],
    [250, 50],
    [350, 100],
    [300, 200],
    [200, 200]
], np.int32)

points = points.reshape((-1, 1, 2))
cv2.fillPoly(
    image,
    [points],
    (0, 255, 0)      
)

image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(10, 6))
plt.imshow(image_rgb)
plt.axis("off")
plt.show()
```

# Arithmetic Operations on Images

OpenCV allows us to perform arithmetic operations on images.

In this section, we will learn:

- `cv2.add()` → adds two images
- `cv2.subtract()` → subtracts one image from another

For these operations, both images should have the same size.

Image Addition

```python
import cv2
import matplotlib.pyplot as plt

# Read two images
image1 = cv2.imread("images\images (2).jpg")
image2 = cv2.imread("images\images (3).jpg")

# Check if images were loaded
if image1 is None or image2 is None:
    print("One or both images were not found!")
else:
    # Resize image2 to match image1
    image2 = cv2.resize(image2, (image1.shape[1], image1.shape[0]))

    # Add the two images
    added_image = cv2.add(image1, image2)

    # Convert BGR to RGB
    image1_rgb = cv2.cvtColor(image1, cv2.COLOR_BGR2RGB)
    image2_rgb = cv2.cvtColor(image2, cv2.COLOR_BGR2RGB)
    added_rgb = cv2.cvtColor(added_image, cv2.COLOR_BGR2RGB)

    # Display
    plt.figure(figsize=(15, 5))

    plt.subplot(1, 3, 1)
    plt.imshow(image1_rgb)
    plt.title("Image 1")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(image2_rgb)
    plt.title("Image 2")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(added_rgb)
    plt.title("Added Image")
    plt.axis("off")

    plt.show()
```

## Image Subtraction

`cv2.subtract()` subtracts the pixel values of one image from another.

Syntax:

cv2.subtract(image1, image2)

```python
import cv2
import matplotlib.pyplot as plt

# Read two images
image1 = cv2.imread("images\images (2).jpg")
image2 = cv2.imread("images\images (3).jpg")

# Check if images were loaded
if image1 is None or image2 is None:
    print("One or both images were not found!")
else:
    # Resize image2 to match image1
    image2 = cv2.resize(image2, (image1.shape[1], image1.shape[0]))

    # Subtract image2 from image1
    subtracted_image = cv2.subtract(image1, image2)

    # Convert BGR to RGB
    image1_rgb = cv2.cvtColor(image1, cv2.COLOR_BGR2RGB)
    image2_rgb = cv2.cvtColor(image2, cv2.COLOR_BGR2RGB)
    subtracted_rgb = cv2.cvtColor(subtracted_image, cv2.COLOR_BGR2RGB)

    # Display
    plt.figure(figsize=(15, 5))

    plt.subplot(1, 3, 1)
    plt.imshow(image1_rgb)
    plt.title("Image 1")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(image2_rgb)
    plt.title("Image 2")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(subtracted_rgb)
    plt.title("Subtracted Image")
    plt.axis("off")

    plt.show()
```
